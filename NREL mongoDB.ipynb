```python
from pymongo import MongoClient
import pandas as pd
```

#### Connect to MongoDB

```python
# Connect to MongoDB
client = MongoClient("mongodb://localhost:27017/")
```

```python
# Select the database iris_db
db = client['testing']
```

```python
# Select the database retail_db
collection = db['nrel']
```

```python
# Retrieve the first three records
result = collection.find().limit(3)
# Print the records
for record in result:
    print(record)
```

```python
# Total number of species in this dataset
states_list = collection.distinct("state")
print("States:", states_list)
print("Total number of States:", len(states_list))
```

```python
#show all rows has state "California"
California_records = collection.find({"state": "California"})

print("All records with state 'California':")
for doc in California_records:
    print(doc)
```
